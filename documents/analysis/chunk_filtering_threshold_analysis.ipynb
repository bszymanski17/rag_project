```python
import sys 
import os
current_dir = os.getcwd()
if os.path.basename(current_dir) == "experiments":
    os.chdir("..")
    
project_root = os.getcwd()

if project_root not in sys.path:
    sys.path.append(project_root)

from src.utils.load_settings import load_yaml
from src.retrieval.retrieval_orchestrator import run_retrieval


config = load_yaml.__wrapped__("config/config.yaml")
```

```python
def show_chunks(prompt, top=config["retriever"]["top_k"], threshold=config["retriever"]["distance_treshold"]):
    context, raw_chunks, score = run_retrieval(user_query=prompt, emb_model=config["models"]["embedding_model"], db_path="chroma_db_old", retriever_top_k=top, distance_threshold=threshold)

    print(f"Znaleziono chunków: {len(raw_chunks)}\n")
    for i, doc in enumerate(raw_chunks):
        print(f"CHUNK {i+1} | Score {score[i]}")
        print("-" * 60)
        print(doc)
        print("=" * 60)
```

```python
show_chunks("What is the official title of the financial report for IFC's 2024 fiscal year?")
# 0.194
```

```python
show_chunks("What was IFC's Net Income for the fiscal year ending June 30, 2024?")
# 0.174
```

```python
show_chunks("What is the stated mission of the International Finance Corporation?")
# 0.2795
```

```python
show_chunks("What was the revised subscription deadline for the General Capital Increase (GCI) following the April 2024 extension?")
# 0.2913
```

```python
show_chunks("With which accounting principles do IFC's consolidated financial statements conform?")
# 0.2175
```
